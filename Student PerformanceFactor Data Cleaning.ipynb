```python
#filling the missing values using the mode, which is the most frequently occurring category in each column.
# I will use these categories to fill the missing cells. This preserves your student records instead of deleting rows.
```

```python
# Fill missing values using the mode

df["Teacher_Quality"] = df["Teacher_Quality"].fillna(df["Teacher_Quality"].mode()[0])

df["Parental_Education_Level"] = df["Parental_Education_Level"].fillna(df["Parental_Education_Level"].mode()[0])

df["Distance_from_Home"] = df["Distance_from_Home"].fillna(df["Distance_from_Home"].mode()[0])

print("Missing values filled successfully.")
```

```python
#Checking for statistical outliers.After investigating the score range, i can use the interquartile range (IQR) method to flag unusual numerical
#values.
```

```python
# Using the IQR method to flag potential outliers
numeric_columns = df.select_dtypes(include="number").columns

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) |(df[col] > upper_bound)]

    print(f"{col}: {len(outliers)} potential outliers")
```

```python
#Checking the actual outlier values by inspect the unusual values in each column.
```

```python
# Calculating and display the outlier boundaries and values

for col in ["Hours_Studied", "Tutoring_Sessions", "Exam_Score"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df.loc[(df[col] < lower_bound) | (df[col] > upper_bound),col]

    print(f"\n{col}")
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)
    print("Outlier values:")
    print(outliers.value_counts().sort_index())
```

```python
#Correcting the invalid examination score of 101
```

```python
# Replace scores outside the valid 0-100 range with missing values
df.loc[
    (df["Exam_Score"] < 0) | (df["Exam_Score"] > 100),"Exam_Score"] = np.nan

# Fill invalid or missing exam scores with the median of valid scores
df["Exam_Score"] = df["Exam_Score"].fillna(df["Exam_Score"].median())

# Verify the correction
print("Minimum exam score:", df["Exam_Score"].min())
print("Maximum exam score:", df["Exam_Score"].max())
print("Missing exam scores:", df["Exam_Score"].isnull().sum())
```

```python
#Final data-quality check

print("Dataset shape:", df.shape)
print("\nTotal missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Exam scores outside 0-100:",
      ((df["Exam_Score"] < 0) | (df["Exam_Score"] > 100)).sum())
```
